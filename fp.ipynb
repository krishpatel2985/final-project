```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("2015.csv")
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
top10 = df.sort_values("Happiness Score", ascending=True).head(10)
plt.figure(figsize=(10,6))
sns.barplot(data=top10, x = "Happiness Score", y="Country")
plt.title("Happiest Countries")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="Economy (GDP per Capita)",
    y="Happiness Score"
)
plt.title("GDP vs Happiness score")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="Family",
    y="Happiness Score"
)
plt.title("Family vs Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="Health (Life Expectancy)",
    y="Happiness Score"
)
plt.title("Health VS Happiness score")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="Freedom",
    y="Happiness Score"
)
plt.title("Freedom vs Happiness score")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x="Trust (Government Corruption)",
    y="Happiness Score"
)
plt.title("Government Trust vs Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df["Happiness Score"], bins=20, kde=True)
plt.title("Distribution of Happiness Scores")
plt.show()
```

```python
columns = [
    "Happiness Score",
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom"
]
sns.pairplot(df[columns])
plt.show()
```

```python
top_gdp = df.sort_values("Economy (GDP per Capita)", ascending=False).head(20)

plt.figure(figsize=(12,6))
sns.barplot(data=top_gdp,
            x="Economy (GDP per Capita)",
            y="Country",
            palette="viridis")

plt.title("Top 20 Countries by GDP per Capita")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.regplot(
    data=df,
    x="Freedom",
    y="Happiness Score",
    scatter_kws={"alpha":0.7}
)

plt.title("Freedom vs Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(6,6))

sns.boxplot(
    y=df["Economy (GDP per Capita)"],
    color="orange"
)

plt.title("Boxplot of GDP per Capita")
plt.show()
```
